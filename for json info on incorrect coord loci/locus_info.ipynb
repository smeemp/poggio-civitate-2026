```python
## Imports ##
import pandas as pd
import json
```

### Template

```python
template = {
    "uri" : "",
    "raw" : "",
    "convertedLongitude" : 0,
    "convertedLatitude" : 0
}

# Stores info for all loci, exported as JSON
all_loci = []
```

### Do The Thing (TM)

```python
print("Type \"x\" to end.")

while True:
    try:
        new_locus = template.copy()

        # URI entry
        print("-NEW LOCUS-")
        uri = input("Please enter a locus URI: ")
        if uri == "x":
            raise Exception()
        new_locus["uri"] = uri

        # Coords entry
        while True:
            raw = input("Please enter the raw coords of a new point in the format \"{PointLabel}: {NScoord}/{EWcoord}\": ")
            if raw == "x":
                break
            new_locus["raw"] = raw

            new_locus["convertedLongitude"] = float(input(f"Please enter the converted longitude of this point: "))
            new_locus["convertedLatitude"] = float(input(f"Please enter the converted latitude of this point: "))

            all_loci.append(new_locus.copy())
        
    except:
        print("Process finished.")
        print(all_loci)
        break

            
```
