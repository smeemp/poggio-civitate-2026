### A lot of the PC numbers and links are wrong/typo'd. I'm going to put their sheet locations into a new column to put into mine so we can fix them at the source as well as in my sheet.

```python
import os
import pandas as pd
```

### Mapping PC numbers to locations

```python
pc_map = {}

# Iterate through all .xlsx files
for file in os.listdir("ScaffeleInventory"):

    # Double checking file name
    print(file)

    # In a try-except in case some secret stupid file appears (I'm looking at you .DS_Store)
    s = pd.read_excel(f"ScaffeleInventory/{file}", engine="openpyxl", sheet_name=None)

    try:
        # Each sheet has multiple contained, so add them individually
        for sheet_name, df in s.items():
            for i in range(len(df)):
                pc_map.update({f"{df["Object Number"][i]}" : f"{file}/{sheet_name}"})

    except: 
        print("ruh roh")

print(pc_map)
```

```python
data = pd.read_csv("InventoryWithImageLinks.csv")
data = data.assign(location = [""]*len(data))

for i in range(len(data)):
    data.loc[i, "location"] = pc_map.get(data["object_number"][i])
```

```python
data.to_csv("InventoryWithLocs.csv")
```
