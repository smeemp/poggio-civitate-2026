```python
## IMPORTS ##
import pandas as pd
import os
import shutil
```

### Populating Image Links in Original Spreadsheet

```python
links = pd.read_csv("InventoryWithImageLinks.csv")

links = links.drop(columns="Unnamed: 0")
```

```python
# Will store every imported sheet
sheets = []

# Iterate through all .xlsx files
### FOR EACH SCAFFELE
for file in os.listdir("./ScaffeleInventory"):

    try:
        # # Make new folder for each scaffele if doesn't already exist
        # current_folder = f"./InventoryWithImgs/{file}"
        # if os.path.exists(current_folder):
        #     shutil.rmtree(current_folder)
        # os.mkdir(current_folder)

        s = pd.read_excel(f"ScaffeleInventory/{file}", engine="openpyxl", sheet_name=None)

        # Each sheet has multiple contained, so add them individually
        ### FOR EACH CASSETTA
        with pd.ExcelWriter(f"./InventoryWithImgs/{file}", engine="openpyxl") as writer:
            for sheet_name, df in s.items():
                # Add new column for image links
                df = df.assign(image_link = [""]*len(df))

                # Loop through all elements in this dataframe and find the first instance
                # of matching pc num. Insert that into sheet
                ### FOR EACH OBJECT
                for i in range(len(df)):
                    target = df["Object Number"][i]
                    found_index = -1

                    # Find matching PC number row in sheet w/ links
                    for l in range(len(links)):
                        if links["object_number"][l] == target:
                            found_index = l
                            break
                    
                    # If found, insert link
                    if found_index >= 0:
                        df.loc[i, "image_link"] = links["image_link"][found_index]
                        print("success")

                # Export to file
                df.to_excel(writer, sheet_name=f"{sheet_name}", index=False)

            
    except:
        print(f"Encountered exception. Filename: {file}")
```
