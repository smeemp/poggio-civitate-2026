```python
## IMPORTS ##
import pandas as pd
import os
import random
from time import sleep
import requests
import json
```

### Reading .xslx Files

```python
# Will store every imported sheet
sheets = []

# Iterate through all .xlsx files
for file in os.listdir("ScaffeleInventory"):

    # Double checking file name
    print(file)

    # In a try-except in case some secret stupid file appears (I'm looking at you .DS_Store)
    try:
        s = pd.read_excel(f"ScaffeleInventory/{file}", engine="openpyxl", sheet_name=None)

        # Each sheet has multiple contained, so add them individually
        for sheet_name, df in s.items():
                sheets.append(df)
    except:
        print(f"Encountered exception. Filename: {file}")
```

### Culling Empty Sheets For Efficiency

```python
# Iterate through every sheet and remove empty ones
# Lowkey I think this skips some because of indexing but it's not like essential or anything
for i in range(len(sheets)):
    if i == len(sheets):
        break
    if sheets[i].empty:
        sheets.pop(i)
```

### Merging Into One DataFrame (for ease of looping)

```python
# Merge all sheets together vertically
df = pd.concat(sheets, ignore_index=True)

# Drop all empty rows and unnecessary/weird columns
df.dropna(how="all", inplace=True)
df = df.drop(df.columns[[n for n in range(2,15)]], axis=1)

# Rename columns to be more Python-friendly
df = df.rename(columns={"Object Number" : "object_number", "Open Context Link" : "url"})
```

### Removing Blank PC Numbers (who tf did that)

```python
# Hatred.
df = df[df.object_number != "PC "]
df = df[df.object_number != "PC"]
df = df[df.object_number != " "]
df.reindex()
```

### INITIAL LOOP - Putting T/F if has URL

```python
# Insert initially empty row
df.insert(2, "valid_link", [False]*len(df))

# Verify each link, so we don't try to open broken ones
for i in range(len(df)):
    try:
        if "https://opencontext.org/subjects/" in str(df["url"][i]):
            df.loc[i, "valid_link"] = True
    except:
        continue
```
